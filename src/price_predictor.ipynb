```python
import pandas as pd
from sklearn.model_selection import train_test_split
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

## Reading data

```python
df = pd.read_csv('data/house-prices.csv')

X = df[['SqFt', 'Bedrooms', 'Bathrooms', 'Offers', 'Brick', 'Neighborhood']]
y = df['Price']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.1, random_state=42)
```

```python
#fill the null/NaN value with median in numeric columns
#cause midian is better than mean with outlier data

columns_with_nulls = ['SqFt', 'Bedrooms', 'Bathrooms', 'Offers']

for col in columns_with_nulls:
    X_train[col] = X_train[col].fillna(X_train[col].median())
    X_test[col] = X_test[col].fillna(X_train[col].median())
    df[col] = df[col].fillna(df[col].median())
```

```python
#converting categorical columns to numeric format

df['Brick'] = df['Brick'].map({'No': 0, 'Yes': 1})

#One-Hot encoding
df = pd.get_dummies(df, columns=['Neighborhood'], drop_first=False)
#for converting bool to int
df *= 1
#One-Hot encoding
X_train['Brick'] = X_train['Brick'].map({'No': 0, 'Yes': 1})
X_test['Brick'] = X_test['Brick'].map({'No': 0, 'Yes': 1})

X_train = pd.get_dummies(X_train, columns=['Neighborhood'], drop_first=False)
X_test = pd.get_dummies(X_test, columns=['Neighborhood'], drop_first=False)

#Aligning Columns
#Handles missing categories in test data by initializing their columns with 0
X_train, X_test = X_train.align(X_test, join='left', axis=1, fill_value=0)
#for converting bool to int
X_train = X_train.astype(float)
X_test = X_test.astype(float)

X_test.info(), X_train.info()
```

```python
#check for outlier data with IQR method
# Outlier detection is focused on continuous features (Price, SqFt).
# Discrete variables (Bedrooms, Bathrooms, Offers) have a limited range,
# and categorical ones (Brick, Neighborhood) lack numerical distance,
# making traditional outlier methods like IQR inapplicable to them.
# Even after One-Hot Encoding, outlier detection is not applied to categorical 
# features because they represent binary states (0 or 1). There are no 
# "extreme" values in binary data, only frequency differences.

Q1 = X_train['SqFt'].quantile(0.25)
Q3 = X_train['SqFt'].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = X_train[(X_train['SqFt'] < lower_bound) | (X_train['SqFt'] > upper_bound)]
print(f" number of outliers in {'SqFt'}: {len(outliers)}")
X_train['SqFt'] = X_train['SqFt'].clip(lower_bound, upper_bound)

Q1 = y_train.quantile(0.25)
Q3 = y_train.quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
outliers = y_train[(y_train < lower_bound) | (y_train > upper_bound)]
print(f" number of outliers in {'Price'}: {len(outliers)}")
y_train = y_train.clip(lower_bound, upper_bound)
```

## Model Implementation and Training

```python
#Data scaling/normalizing
    
def standardize(X_train, X_test):
    X_mean = np.mean(X_train, axis=0)
    X_std = np.std(X_train, axis=0)
    #prevent divide by 0
    X_std = np.where(X_std == 0, 1e-8, X_std)
    
    X_train_scaled = (X_train - X_mean) / X_std
    X_test_scaled = (X_test - X_mean) / X_std

    return X_train_scaled, X_test_scaled

X_train_scaled, X_test_scaled = standardize(X_train, X_test)

X_train_np = X_train_scaled.values
y_train_np = y_train.values.reshape(-1, 1)
X_test_np = X_test_scaled.values
y_test_np = y_test.values.reshape(-1, 1)

#add bias
X_train_np = np.c_[np.ones((X_train_np.shape[0], 1)), X_train_np]
X_test_np = np.c_[np.ones((X_test_np.shape[0], 1)), X_test_np]
```

```python
class GDLinearRegression:
    def __init__(self, learning_rate, epochs):
        self.learning_rate = learning_rate
        self.epochs = epochs
        self.weights = None
        self.rmse_history = []

    def compute_rmse(self, X, y):
        predictions = np.dot(X, self.weights)
        mse = np.mean((predictions - y) ** 2)
        rmse = np.sqrt(mse)
        return rmse
    
    def fit(self, X, y):
        n_samples, n_features = X.shape
        # Initialize weights with zero value
        self.weights = np.zeros((n_features, 1))

        for i in range(self.epochs):
            y_pred = np.dot(X, self.weights)
            error = y_pred - y
            cur_rmse = self.compute_rmse(X, y)

            # Compute the direct gradient of RMSE
            # To avoid division by zero when RMSE becomes zero, add a very small number.
            gradient = (1 / (n_samples * (cur_rmse + 1e-15))) * np.dot(X.T, error)
            self.weights -= self.learning_rate * gradient

            self.rmse_history.append(cur_rmse)
    
    def predict(self, X):
        return np.dot(X, self.weights)  
```

```python
# Create the model and train it

model = GDLinearRegression(learning_rate=1, epochs=200000)
model.fit(X_train_np, y_train_np)

# Plot the convergence graph
plt.figure(figsize=(8, 5))
plt.plot(range(len(model.rmse_history)), model.rmse_history, color='blue')
plt.xlabel('epochs')
plt.ylabel('RMSE')
plt.title('Training RMSE history')
plt.grid(True)
plt.show()
```

```python
# predict on test
y_pred_test = model.predict(X_test_np)

# calculate final errors
train_rmse = model.compute_rmse(X_train_np, y_train_np)
test_rmse = model.compute_rmse(X_test_np, y_test_np)

print(f"Final Training RMSE: {train_rmse:.2f}")
print(f"Final Testing RMSE: {test_rmse:.2f}")

# Compare the first few predictions with actual values in the test set
comparison = pd.DataFrame({
    'Actual Price': y_test_np.flatten(),
    'Predicted Price': y_pred_test.flatten()
})
print("Comparison of actual and predicted values:\n")
print(comparison.head())
```
